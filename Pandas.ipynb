#Creating Dataframe

```python
import pandas as pd
import numpy as np
```

```python
student_data = [
    [100,40,10],
    [90,70,7],
    [60,50,6],
    [40,60,5]
]

pd.DataFrame(student_data,columns =['marks','iq','percentage'])
```

```python
student_dict = {
    'name':['bhalu','kalu','lalu'],
    'marks':[50,60,70],
    'iq':[20,40,50],
    'percentage':[6,7,8],
    'package':[10,7,8]
    }
data = pd.DataFrame(student_dict)
data.set_index('name',inplace=True)
data
```
