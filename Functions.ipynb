#def functtion name(parameters):

```python
def is_even(num):
    """
    This Function returns if a given no is odd or even.
    input any valid integer 
    output - odd or even 
    """
    if type(num)==int:
        if num%2==0:
            return 'even'
        else:
            return 'odd'
    else:
        return 'Invalid datatype'
    
```

```python
for i in range(1,11):
    x = is_even(i)
    print(x)
```

```python
is_even('Manish')
```

#*args (non keyword arguments)

```python
def multiply(a,b,c):
    return a*b*c
```

```python
multiply(2,3)
```

```python
multiply(1,2,3)
```

```python
def multiply(*args): 
    product = 1
    
    for i in args:
        product = product*i
    return product
```

```python
multiply(1,2,3,4,5)
```

```python
def multiply(*Man):  # no issue with args name change 
    product = 1
    
    for i in Man:
        product = product*i
    return product    
```

```python
multiply(2,3,4,5,76,7,8,9,90)
```
