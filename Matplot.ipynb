```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
data = pd.read_csv('sharma-kohli.csv')
data
```

```python
plt.plot(data['index'],data['V Kohli'])
plt.plot(data['index'],data['RG Sharma'])
```

```python
plt.plot(data['index'],data['V Kohli'])
plt.plot(data['index'],data['RG Sharma'])
plt.title("Rohit sharma vs Virat kholi stats")
plt.xlabel('Season')
plt.ylabel('Run')
```

```python
plt.plot(data['index'],data['V Kohli'],color = 'orange',linestyle = 'dashed',linewidth = 3,marker = '.',markersize = 10,label = 'Virat')
plt.plot(data['index'],data['RG Sharma'],color = 'green',linestyle = 'dashdot',linewidth = 2,marker = '.',markersize = 10,label = 'Rohit_S')
plt.title("Rohit sharma vs Virat kholi stats")
plt.xlabel('Season')
plt.ylabel('Run')
plt.legend()
```

```python
plt.plot(data['index'],data['V Kohli'],color = 'orange',linestyle = 'dashed',linewidth = 3,marker = '.',markersize = 10,label = 'Virat')
plt.plot(data['index'],data['RG Sharma'],color = 'green',linestyle = 'dashdot',linewidth = 2,marker = '.',markersize = 10,label = 'Rohit_S')
plt.title("Rohit sharma vs Virat kholi stats")
plt.xlabel('Season')
plt.ylabel('Run')
plt.legend()
plt.grid()
plt.show()
```
