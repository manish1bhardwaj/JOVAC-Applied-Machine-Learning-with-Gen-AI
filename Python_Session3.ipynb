#List Comprehension

```python
#Add 1 to 10 numbers in a List By for loop
L = []
for i in range(1,11):
    L.append(i)
print(L)
```

```python
#Add 1 to 10 numbers in a List By List Comprehension
[i for i in range(1,11)]
```

```python
L = [1,2,3,4,5,6]
[i**2 for i in L]
```

```python
#print all the Number divisible by 5 in the range of 1 to 50
[i for i in range(1,51) if i%5==0]
```

```python
#find language which start with letter P
languages = ['java','python','php','C','javascript']
i for i in languages if i[0]=='p'
[i for i in languages if i.startswith('p') ]
```
